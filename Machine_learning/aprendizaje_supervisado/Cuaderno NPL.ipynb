## *ELIMINACIÓN DE PALABRAS "VACIAS" EN EL MENSAJE:

```python
import nltk
from nltk.corpus import stopwords
nltk.download("stopwords")
nltk.download('punkt')
```

```python
spanish_stopwords = set(stopwords.words("spanish"))
filtered_words = [word for word in word_tokenize(text) if word.lower() not in spanish_stopwords]
print(filtered_words)
```
