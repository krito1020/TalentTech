```python
import es_core_news_sm
nlp = es_core_news_sm.load()
```

```python
# Funcion para preprocesar los datos de texto

def preprocess(text):
    #Eliminar Stoprwords:
    doc = nlp(text)
    filtered_tokens = []
    for token in doc:
        if token.is_stop or token.is_punct:
            continue 
        filtered_tokens.append(token.lemma_)                # Agrega el texto lemalizado  
    return ' '.join(filtered_tokens)
```

```python
df['texto preprocesado'] = df['text'].apply(preprocess)
```
